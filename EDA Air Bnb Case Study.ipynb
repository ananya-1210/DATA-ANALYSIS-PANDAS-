```python
def outlier_detection(df):
    Uo = df['75%']+(df['75%']-df['25%'])*1.5
    Lo = df['25%']-(df['75%']-df['25%'])*1.5

    df['UO'] = df['max']>Uo
    df['LO'] = df['min']<Lo
    df['Outiler'] = df['UO']|df['LO']
    
    return df
```
